```python
import json
import pandas as pd

file_path = "data/raw/historial-de-reproducciones.json"

with open(file_path, "r", encoding="utf-8") as file:
    data = json.load(file)

df = pd.DataFrame(data)

df.head()
```

```python
def extraer_canal(subtitles):
    if not isinstance(subtitles, list):
        return None
    if len(subtitles) > 0:
        return subtitles[0].get("name")
    return None

df["canal"] = df["subtitles"].apply(extraer_canal)

df[["title","canal"]].head()
```

```python
df["canal"].value_counts().head(10)
```

```python
pd.to_datetime(df["time"], errors="coerce")
```

```python
df["time"] = pd.to_datetime(df["time"], errors="coerce")
df["time"].head()
```

```python
df = df.dropna(subset=["time"])
df.shape
```

```python
df.columns
```

```python
df["año"] = df["time"].dt.year
df["mes"] = df["time"].dt.month
df["dia"] = df["time"].dt.day
df["hora"] = df["time"].dt.hour
df["dia_semana"] = df["time"].dt.day_name()
```

```python
df.columns
```

```python
df["hora"].value_counts().sort_index()
```

```python

def clasificar_periodo(hora):
    if hora <= 5:
        return "madrugada"
    elif hora <= 11:
        return "mañana"
    elif hora <= 17:
        return "tarde"
    else:
        return "noche"
```

```python
df["periodo"] = df["hora"].apply(clasificar_periodo)
df["periodo"].value_counts()
```

```python
df_madrugada = df[df["periodo"] == "madrugada"]
```

```python
df_madrugada["canal"].value_counts().head(10)
```

```python
df_bad_bunny = df[df["canal"] == "Bad Bunny"]
```
